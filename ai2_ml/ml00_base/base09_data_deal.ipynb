```python
import numpy as np
import pandas as pd

data = np.array([
    [50, 200],
    [60, 250],
    [80, 300],
    [100, 400]
])
df = pd.DataFrame(data, columns=["size", "price"])
print(df)
```

```python
# 手动实现 归一化
size_min = np.min(df["size"])
size_max = np.max(df["size"])
size_ = np.array([(df['size'] - size_min) / (size_max - size_min)])
print(size_, size_.shape)
# df['size_'] = size_.reshape(-1, 1)
# print(df)

x_min = df.min()
x_max = df.max()
df_norm = (df - x_min) / (x_max - x_min)
print("手动归一化：\n", df_norm)
```

```python
# 手动标准化
size_mean = df['size'].mean()
size_std = df['size'].std()
size__ = np.array([(df['size'] - size_mean) / size_std])
print(size__, size__.shape)

df_mean = df.mean()
df_std = df.std()
df__ = (df - df_mean) / df_std
print("手动标准化：\n", df__)
```

```python
# 使用skleam实现

from sklearn.preprocessing import MinMaxScaler

df_norm2 = MinMaxScaler().fit_transform(df)
print("sklearn归一化：\n", df_norm2)

from sklearn.preprocessing import StandardScaler

df_std2 = StandardScaler().fit_transform(df)
print("sklearn标准化：\n", df_std2)

```
