基本特征处理方法
1. 归一化，标准化
2. 将数据集里面所有的 '?' 数据全部去除
3. 对数据进行独热编码



```python
import numpy as np
import pandas as pd

data = np.array([
    [50, 200],
    [60, 250],
    [80, 300],
    [100, 400]
])
df = pd.DataFrame(data, columns=["size", "price"])
print(df)

# 手动实现 归一化
size_min = np.min(df["size"])
size_max = np.max(df["size"])
size_ = np.array([(df['size'] - size_min) / (size_max - size_min)])
print(size_, size_.shape)
# df['size_'] = size_.reshape(-1, 1)
# print(df)

x_min = df.min()
x_max = df.max()
df_norm = (df - x_min) / (x_max - x_min)
print("手动归一化：\n", df_norm)


# 手动标准化
size_mean = df['size'].mean()
size_std = df['size'].std()
size__ = np.array([(df['size'] - size_mean) / size_std])
print(size__, size__.shape)

df_mean = df.mean()
df_std = df.std()
df__ = (df - df_mean) / df_std
print("手动标准化：\n", df__)


# 使用skleam实现

from sklearn.preprocessing import MinMaxScaler

df_norm2 = MinMaxScaler().fit_transform(df)
print("sklearn归一化：\n", df_norm2)

from sklearn.preprocessing import StandardScaler

df_std2 = StandardScaler().fit_transform(df)
print("sklearn标准化：\n", df_std2)

```

```python

# 2、快速对数据进行标准化，归一化
from sklearn.preprocessing import StandardScaler
X_train = []
X_test = []
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
# 为什么X_train调用fit_transform，而X_test却调用transform？
# 机器学习中数据标准化（或归一化）最核心的一个原则 —— 只在训练集上学习参数，然后 用同样的参数作用于测试集
# 训练集 fit_transform() 学习均值和标准差，并进行标准化
# 测试集 transform() 使用训练集的参数进行标准化，避免数据泄露
```

```python
# 3、将数据集里面所有的 '?' 数据全部去除 ,    去除?数据

def feature_deal_question_mark(df: pd.DataFrame):
    # 遍历每个元素，如果该行存在'?'，则删除该行

    # 将 '?' 替换为 np.nan
    df = df.replace('?', pd.NA)

    # 删除含有缺失值（NaN）的行
    df = df.dropna()

    return df
```

```python

# 5、特征数据读热编码
heart_disease = pd.DataFrame()
dropFirstColumn = True
# 胸痛类型：4种分类（名义变量），cp：0-典型心绞痛，1-非典型心绞痛，2-非心绞痛，3-无症状，cp1,cp2,cp3,cp4每一个都对应0，1
# 将原来用取值0,1,2,3的一个特征转化为现在用四个特征来分别表示有没有，即将一个特征独热编码为4个特征

# 处理 cp
cp1 = np.array([1 if cp == 0 else 0 for cp in heart_disease['cp']])
cp2 = np.array([1 if cp == 1 else 0 for cp in heart_disease['cp']])
cp3 = np.array([1 if cp == 2 else 0 for cp in heart_disease['cp']])
cp4 = np.array([1 if cp == 3 else 0 for cp in heart_disease['cp']])
if not dropFirstColumn:
    heart_disease['cp1'] = cp1
heart_disease['cp2'] = cp2
heart_disease['cp3'] = cp3
heart_disease['cp4'] = cp4
heart_disease = heart_disease.drop('cp', axis=1)

# 手动独热编码冗长，容易出错, 可使用 pandas.get_dummies 一行解决，自动处理多列
feature_deal_heart_disease = pd.get_dummies(
    heart_disease,
    columns=['cp', 'restecg', 'slope', 'thal'],
    drop_first=True  # 避免多重共线性
)
```
