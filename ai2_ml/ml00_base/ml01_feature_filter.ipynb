```python
# 1）低方差过滤法
# 对于特征的选择，可以直接基于方差来判断，这是最简单的。低方差的特征意味着该特征的所有样本值几乎相同，对预测影响极小，可以将其去掉。
```

```python
import numpy as np
```

```python
# 特征选择方法
# 低方差过滤

a = np.random.randn(100)  # 随机生成100个数据，a作为特征1

b = np.random.normal(5, 0.1, 100)  # 随机生成100个方差为0.01的数据，作为特征2

c = np.random.normal(5, 0.005, 100)  # 随机生成100个方差为0.01的数据，作为特征2

print(a.shape, b.shape, c.shape)

# 将a,b,c三个特征作为所有特征，进行特征组合
# np.vstack(a,b,c).shape=(3,100)这个不符合100个样本点3个特征的数据要求，因此对齐进行转置,
# 即可得到符合要求的特征数据X.shape=(100,3)
X = np.vstack((a,b,c)).T
print(X, X.shape)

# 经过这样处理之后，X就有100个样本每个样本有3个特征，其中第三个特征的方差小于0.01，因此在下面的特征方差过滤中会被移除掉，只保留前面两个特征。这个就是本次要演示的特征过滤方法。
# 此外，方差越小说明特征之间的差别就越小，也就是说每一个特征都差不多，即这个特征用来决定标签的作用也就不大了，所以可以移除。


# 特征数据准备好了
```

```python
# 现在使用form sklearn.feature_selection import VarianceThreshold进行特征方差过滤
from sklearn.feature_selection import VarianceThreshold

vt = VarianceThreshold(threshold=0.01)
x_transfored = vt.fit_transform(X)

print(x_transfored, x_transfored.shape)
# 输出 (100, 2)，那么也就是说方差小雨0.01的c特征已经被过滤了
```
