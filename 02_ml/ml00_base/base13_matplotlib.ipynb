```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# 定义函数y=sinx
x = np.arange(0, 6, 0.1)
y = np.sin(x)
# 画出函数图像
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sin(x)")
plt.show()
```
