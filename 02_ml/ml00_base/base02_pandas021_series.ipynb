```python
# 索引属性
s = pd.Series([1,2,3,4,5], index=['a', 'b', 'c', 'd', 'e'], name='数据标签, 描述这一列的数据是什么数据')

# loc , iloc
# loc表示按显示索引，即由指定index属性的时候才能勇敢， loc -->>> location
# iloc表示按从0开始的数字索引进行访问；iloc -->>> integer location
# loc和iloc可以进行切片操作
print(s.loc['a'])
# 注意：：：loc的标签索引切片是包含条件两边的，即右侧也是闭合的
print(s.loc['a':'a'])
print(s.loc['a':'b'])
print(s.loc['a':'c'])
print()
print(s.iloc[0])
# iloc的数字切片是前闭后开的，即不包含条件右侧
print(s.iloc[0:0])
print(s.iloc[0:1])
print(s.iloc[0:2])
```

```python
# 索引属性，索引单个数据
s = pd.Series([1,2,3,4,5], index=['a', 'b', 'c', 'd', 'e'], name='数据标签, 描述这一列的数据是什么数据')

# at, iat
# Series.at：按「标签」取单个值（label）
# Series.iat：按「位置」取单个值（integer position）
# at访问数据比loc访问数据更快
print(s.at['b'])
# at也可以修改数据
s.at['b'] = 666
print(s.at['b'])
# 如果尝试访问一个不存在的key，会抛出错误： KeyError: 'aaa'
# print(s.at['aaa'])
print(s.iat[2])
# IndexError: index 123 is out of bounds for axis 0 with size 5
# print(s.iat[123])

# at或iat的使用场景：1、在循环里频繁访问单个元素；2、明确只取一个值
```

```python

# 注意⚠️⚠️⚠️：loc和at只能使用index指定的数据标签访问，而iloc和iat只能使用从0开始的数字索引访问
# 如下是经典的错误
# 数据是[1,2,3,4,5]，而使用[10,20,30,40,50]来指定数据的标签
serr = pd.Series([1,2,3,4,5], index=[10,20,30,40,50])
# 演示：loc和ta只能使用inde向指定的标签访问
print(serr.loc[10])
print(serr.loc[10:30])
print(serr.at[10])
# print(serr.at[1]) KeyError: 1
print()
# 演示：iloc和iat只能使用从0开始的数字索引访问
print(serr.iloc[0:1])
# IndexError: index 10 is out of bounds for axis 0 with size 5
# print(serr.iat[10])
```

```python

# 某城市一周中每天的最高温度如下
temp = pd.Series([28,31,29,32,30,27,33], index=['周1', '周2', '周3', '周4', '周5', '周6', '周7'])
# 温度超过30的天数
print(temp[temp > 30].count())
print(temp.mean())
# 温度低到高排序
print(temp.sort_values()) # 升序
print(temp.sort_values(ascending=False)) # 降序
# 找出温度变化最大的两天
max = 0
max_index = 0
for i, v in enumerate(temp):
    if i == len(temp)-1:
        break
    # diff = temp[i+1] - v
    diff = temp.iat[i+1] - v
    if abs(diff) > max:
        max = diff
        max_index = i
print(max, max_index)
print(f"温差最大的两天是{max_index+1}和{max_index +2}, 最大温差是{max}读")
print()
# 找相邻两个数据之间的差值
print(temp.diff())
print(temp.diff().abs().sort_values(ascending=False))
print(temp.diff().abs().sort_values(ascending=False).iat[0])
print(temp.diff().abs().sort_values(ascending=False).index)
max_i = temp.diff().abs().sort_values(ascending=False).index[0]
print("标签索引：",max_i)
# 通过标签索引获取数字索引
i_index = temp.index.get_loc(max_i)
print(i_index)
print(i_index-1)
# 通过数字索引获取标签索引
print(temp.index)
print(temp.index[i_index])
# 已知「值」，获取对应的 index（可能多个）
# s[s == 5].index
```

```python
# 某股票连续10天的收盘价
s = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2], index=pd.date_range('2023-01-01',periods=10))
# print(s)
# 计算每日收益率 = 当日收盘价 / 当前日收盘价 - 1
print(s[1:s.size].values)
print(s[0:s.size-1].values)
lst = [s[1:s.size].values[i]/ s[0:s.size-1].values[i] - 1 for i in range(len(s)-1)]
lst = [np.nan] + lst
s0 = pd.Series(lst, index=pd.date_range('2023-01-01',periods=10))
print(s0)
# 找出最大，最低对应的日期
print(s0.max(), s0.idxmax(), s0.index.get_loc(s0.idxmax()))
print(s0.min(), s0.idxmin(), s0.index.get_loc(s0.idxmin()))
# 波动率
print(s0.std())

# pd内置收益率的函数
print(s.pct_change())
```

```python

# 某产品过去12个月的销售量
# freq表示频率，M表示月,S表示每个月的开始第一天，E表示月末最后一天
s = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220], index=pd.date_range('2022-01-01', periods=12, freq='MS'))
print(s)

# 计算季度平均销量, Q表示季度
# 重新采样将每3个月合并在一起
print(s.resample('QS').mean())


# 环比：这个月比上一个月增加了多少
# 同比：今年的某个月比去年的对应那个月增加了多少

# 找出连续增长超过2个月的月份
c = s.pct_change()
print(c)
b = c > 0
print(b)
# 使用滑动窗口计算连续超过两个月的
print(b.rolling(3))
print(b.rolling(3).sum())
print(b.rolling(3).sum() >= 3)
print(b[b.rolling(3).sum() >= 3])
print(b[b.rolling(3).sum() >= 3].keys())
print(b[b.rolling(3).sum() >= 3].keys().tolist())


pass
```

```python
import numpy as np
# 随机生成商店每小时的销售额度
np.random.seed(123)
s = pd.Series(
    np.random.randint(0, 100, 24),
    index=pd.date_range("00:00:00", periods=24, freq="h")
)
print(s)
# 按天重采样计算每天的销售额度
print(s.resample("D").sum())
# 计算每天营业时间销售额  和 非营业时间销售额 的比例
print(s[8:23])
print(s[8:23].sum())
print(s.sum() - s[8:23].sum())
print((s[8:23].sum()) / (s.sum() - s[8:23].sum()))

today_sum = s.sum()
business_data = s[(s.index.hour >= 8) & (s.index.hour <= 22)].sum()
business_data = s.between_time('08:00', '22:00').sum()
# s.between_time()
print(business_data)
not_business_data = s.sum() - business_data
not_business_data = s[(s.index.hour > 22) | (s.index.hour < 8)].sum()
print('not_business_data: ', not_business_data)
print(business_data / not_business_data)

# drop在原始数据里面去除某些数据
print(s.drop(s.between_time('08:00', '22:00').index).sum())

# 超出销售额最高的三个小时
print(s.sort_values(ascending=False))
print(s.sort_values(ascending=False)[0:3])
print(s.sort_values(ascending=False)[0:3].index)

print(s.nlargest(3))
print(s.nsmallest(3))
print()
```
