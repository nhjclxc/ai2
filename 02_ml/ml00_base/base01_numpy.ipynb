```python
# 2.2预定义形状填充，基于全0，全1的生成
print(np.zeros((2, 3)))
print(np.zeros((3, 2)))

print(np.ones((2, 3)))
print(np.ones((3, 3)))

print(np.empty((2, 3)))

# full用具体的某一个数值填充具体形状的arr
print(np.full((2, 3), 2))
# print(np.full())
```

```python
# ndarray的索引与切片
arr = np.array([[1, 2, 3, 4, 5, 6], [11, 12, 13, 14, 15, 16], [21, 22, 23, 24, 25, 26], [31, 32, 33, 34, 35, 36]])
print(arr)
# 获取指定位置的元素
# print(arr[0][0])
# print(arr[0,0])
# print(arr[0][2])
# print(arr[0,2])

# # 获取行切片
# print("获取行切片")
# print(arr[0])
# print(arr[1])
# print(arr[0][1:6])
# print(arr[0][1:6:2])

# 获取列切片
print("获取列切片")
print(arr[0, 1])
print(arr[1, 1])
print(arr[2, 1])
print(arr[3, 1])
print(arr[:, 1])
print(arr[:, 1].reshape((4, 1)))
print(arr[:, 1:3])
# arr[:, 1] 与 arr[:][1]
# arr[:] 表示arr数组的拷贝对象，那么arr[:][1]表示取arr的第2行
print(arr[:])
# 故，在ndarray中，取多维数据都使用[i,j,k]的取法

# bool索引
# arr % 3 == 0 计算所有元素是否符合条件，符合条件的的对应位置为True，否则为False，bool 掩码（mask）
# arr[arr % 3 == 0] 将得到的bool矩阵和原矩阵的每一个位置做bool索引，bool矩阵中为True的元素值将被留下构成新的矩阵返回
print(arr % 3 == 0)
print(arr[arr % 3 == 0])

# 筛选第2行所有符合 arr % 3 == 0 的数据
print(arr[1, arr[1, :] % 3 == 0])

# 筛选第 2 列所有符合 arr % 3 == 0 的数据
print(arr[:, 1])
print(arr[:, 1] % 3 == 0)
print(arr[arr[:, 1] % 3 == 0, 1])
```

```python

# 用函数实现切片
arr = np.array([
    [1, 2, 3, 4, 5, 6],
    [11, 12, 13, 14, 15, 16],
    [21, 22, 23, 24, 25, 26],
    [31, 32, 33, 34, 35, 36]])
print(arr)

# axis 表示“沿着哪一条轴去操作”
# axis=0 → 沿着行的方向（对每一列）
# axis=1 → 沿着列的方向（对每一行）
# axis的含义本质是表示你要操作第几层（多维的时候才有），
#   axis=0表示操作第0层，前面的数组[0, 1, 2]：表示取这个向量每一个第0层中的元素[0, 1, 2]索引对应的元素
#   axis=1表示操作第1层，前面的数组[0, 1, 2]：表示取这个向量每一个第1层中的元素[0, 1, 2]索引对应的元素
print(np.take(arr, [0, 1, 2], axis=0))
print(np.take(arr, [0, 1, 2], axis=1))
```

```python
# 一维ndarray的运算
# 两个进行运算的array的形状必须相同
arr11 = np.array([1, 2, 3])
arr12 = np.array([7, 8, 9])

print(arr11 + 1)
print(arr11 - 2)
print(arr11 * 3)
print(arr11 / 4)
print()

print(arr11 + arr12)
print(arr11 - arr12)
print(arr11 * arr12)
print(arr11 / arr12)

print()

# 二维ndarray的运算
arr21 = np.array([[1, 2], [5, 6]])
arr22 = np.array([[2, 3], [7, 8]])

print(arr21 + arr22)
print(arr21 - arr22)
print(arr21 / arr22)
# 使用 * 表示每一个元素相对应的位置相乘
print(arr21 * arr22)
# 使用 @ 表示进行矩阵乘法
print(arr21 @ arr22)

```

```python
# 基本数学运算
arr = np.array([[2, 3, 1], [5, 5, 6]])
print(np.sqrt(arr))
print(np.sum(arr))
print(np.std(arr))

# 1️⃣ np.percentile —— 分位数（非常常用）
print(np.percentile(arr, 50))  # 中位数
print(np.percentile(arr, [25, 50, 75]))
# 2️⃣ np.greater / np.less —— 显式比较（广播友好）
print(np.greater(arr, 3))  # arr中每个元素与给定数值进行比较，大于给定数的返回True，返回mask矩阵
print(np.less(arr, 5))
# 3️⃣ np.where —— 条件选择（核心）
# where形式1，三元运算
# 以下接口等价:[arr[i] if arr[i] > 3 else 0 for i in range(len(arr))]
print(np.where(arr > 3, arr, 666))  # arr中每个元素进行 > 3比较,true的返回arr中对应的数据,否则返回给定的666
# 形式 2：找索引（重要）
print(np.where(arr > 3))
print(np.where(arr > 3)[0])
# 4️⃣ np.unique —— 去重 + 统计神器
print(np.unique(arr))  # 将原矩阵去除重复后返回不重复的列表
print(np.unique(arr, return_counts=True))  #
print(np.unique(arr, return_inverse=True))  #
# 5️⃣ np.in1d -> np.isin —— 元素是否属于集合（老函数）
print(np.isin(np.array([1, 6, 66]), arr))
# 6️⃣ np.split —— 切数组（不是切片）
print(np.split(arr, 2))
# 7️⃣ np.isnan —— NaN 判断（必须会）
nan_arr = np.array([1, 6, 66, np.nan, 666, np.nan])
print(np.isnan(nan_arr))
print(~np.isnan(nan_arr))  # 是py的按位取反运算符,那么np.isnan(nan_arr)对应的mask矩阵会取反
print(nan_arr[~np.isnan(nan_arr)])  # 基于上一步的mask矩阵,返回不是nan的所有数据
# 8️⃣ np.sort —— 返回排序副本
print(arr)
print(np.sort(arr))  # 注意:sort不修改原矩阵,此外py原生的arr.sort()是就地排序
print(arr)

# 一、np.where —— “找位置”到底在找什么,返回的是：满足条件的“坐标”
arr2 = np.array([1, 5, 6, 9, 2, 3])
print(np.where(arr2 >= 5))  # (array([1, 2, 3]),)
print(np.where(arr2 >= 5)[0])  # 返回[1 2 3]表示的是,arr2中索引为[1 2 3]的位置的数据满足>=5的条件
arr22 = np.array([[3, 5, 12], [6, 8, 9]])
print(np.where(arr22 % 3 == 0))
print(np.where(arr22 % 3 == 0)[0])
print(np.where(arr22 % 3 == 0)[1])
print(arr22[0, 0], arr22[0, 2], arr22[1, 0], arr22[1, 2])

arr3 = np.array([10, 20, 10, 30, 20])
# inverse 表示：原数组中每个元素，在 unique_vals 里的“编号”
unique_vals, inverse = np.unique(arr, return_inverse=True)
print(unique_vals)  # [10, 20, 30] 表示去重后的数组
# inverse 长度与原数组相同, inverse的下标表示原数组的值,对应存的值为原数组的那个值在新数组中的索引
print(inverse)  # [0, 1, 0, 2, 1] 表示 unique_vals数据中每个索引位置元素在原数组 中出现的位置

labels = np.array(["apple", "banana", "apple", "orange"])
classes, encoded = np.unique(labels, return_inverse=True)
print(classes)
print(encoded)
```

```python
arr = np.random.randint(0, 100, 10)
print(arr)
# 不及格 <= 60 < 及格 <= 80 <= 优秀 < 100
print(np.where(arr >= 60, np.where(arr >= 80, '优秀', '及格'), '不及格'))
print(np.where(
    arr < 60,
    '不及格',
    np.where(arr < 80, '及格', '优秀')
))

conditions = [
    arr < 60,
    (arr >= 60) & (arr < 80),
    arr >= 80
]

choices = ['不及格', '及格', '优秀']

result = np.select(conditions, choices, default="unknow")
print(result)
```
