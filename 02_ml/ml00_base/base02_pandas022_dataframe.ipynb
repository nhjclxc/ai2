```python
df = pd.DataFrame(
    {
        'name': ['张三', '里斯', '王五', '赵六', '钱七', '克里斯'],
        'age': [18,25,36,19,59,28],
        'score': [92.6,56.5,82,38.6,61.6,50]
    },
    index=['id0','id1','id2','id3', 'id4', 'id5'],
)
print(df)
print('行索引：', df.index)
print('列标签：', df.columns)
print('值：', df.values)
print(df.values.shape)
print(df.values.size)
print(df.values.dtype)
print(df.ndim)
print(df.shape)
print(df.size)
print(df.dtypes)
print(df.T)
print(df.T.index)
print(df.T.columns)
```

```python

# 显示索引：loc，at
# 隐式索引：iloc，iat
# loc和iloc用于同时获取多个数据，或单个数据
# at和iat用于读取某个元素
# df['列标签'].xxx用于访问多个或单个数据

# loc是根据标签index或column进行选取的
# iloc是根据数字索引选取的
print('1、loc的使用方法')
# 选取某一行
print(df.loc['id0'])
# 选取多行
print(df.loc[['id0', 'id1']])
# 选取某一列
print(df.loc[:, 'name'])
# 选取多列
print(df.loc[:, ['name', 'age']])
# 使用loc选取指定的行列数据
print(df.loc['id0', 'name'])
print(df.loc[['id0', 'id1'], 'name'])
print(df.loc[['id0', 'id1'], ['name', 'age']])


```

```python

print('2、iloc的使用方法')
# 选取某一行
print(df.iloc[0])
# 选取多行
print(df.iloc[[0,1]])
# 选取某一列
print(df.iloc[:, 0])
# 选取多列
print(df.iloc[:, [0,1]])
# 使用loc选取指定的行列数据
print(df.iloc[0,1])
print(df.iloc[[0,1],[0,1]])


print()
# at或iat的第一个参数指的是行的索引，第二个参数指的是列的索引
# at或iat只能用来访问单个元素
# at只能使用指定的标签访问，而iat只能使用数字索引访问
```

```python

print('3、at或iat的使用方法')
print(df.at['id0','age'])
print(df.iat[0,1])
```

```python

print('、扩展获取方法，主要用于获取列数据，常用于获取某列数据进行分析')
# 获取整列
print(df['name'], type(df['name']))
print(df[['name', 'age']] )
print(df.name)
# 获取指定
print(df['name'].loc['id0'])
print(df['name'].iloc[0])
print(df['name'].at['id0'])
```

```python
# bool索引
print(df['score']> 60)
print(df[df['score']> 60])
print(df[(df['score']> 60) & (df.age <30)])
```

```python
# 随机抽样获取数据
print(df.sample())
print(df.sample(2))
print(df.sample(2, axis=0)) # 按行进行抽样
print(df.sample(2, axis=1)) # 按列进行抽样
print(df.sample(2, axis=1, random_state=1)) # 按列进行抽样
```

```python
data = {
    'name': ['张三', '里斯', '王五', '赵六', '钱七'],
    'math': [85,92,78,88,95],
    'english': [90,88,85,92,80],
    'physics': [75,80,88,85,90],
}
df = pd.DataFrame(data)
print(df)
# 计算每位学生的总分和平均分
print(df.iloc[0,0])
print(df.iloc[0,1])
print(df.iloc[0,2])
print(df.iloc[0,3])
print(df.iloc[0,1] + df.iloc[0,2] + df.iloc[0,3])
print(df.iloc[0,1:4])
print(df.iloc[0,1:4].sum())
print(df.iloc[1,1:4].sum())
print(df.iloc[2,1:4].sum())
print(df.iloc[3,1:4].sum())
print(df.iloc[4,1:4].sum())
total = np.array([df.iloc[i,1:4].sum() for i in range(df.shape[0])])
print(total)
df['total'] = total
print(df)
df['mean'] = total / 3
print(df)

# 找出数学成绩高于90分 或 英语成绩高于85分的学生
print(df['math'] > 90)
print(df['english'] > 85)
print((df['math'] > 90 ) | (df['english'] > 85))
print(df[(df['math'] > 90 ) | (df['english'] > 85)])

# 按总分从高到低排序
print(df.sort_values('total', ascending=False))
print(df.sort_values('total', ascending=False).head(3))
print(df.sort_values('total', ascending=False).head(3)['name'])
print(df.sort_values('total', ascending=False).head(3)['name'].tolist())

print(df.iloc[:,1:4].sum(axis=1))
print(df.iloc[:,1:4].sum(axis=1) / 3)
print(df[['math', 'english', 'physics']].sum(axis=1))
print(df[['math', 'english', 'physics']].sum(axis=1) / 3)
print(df.iloc[:,1:4].mean(axis=1))
pass
```

```python

data = {
    'product': ['A', 'B', 'C', 'D'],
    'price': [100,150,200,120],
    'sale': [50,30,20,40]
}
df = pd.DataFrame(data)
print(df)
# 计算每种产品的总销售额, 总销售额 = 单价 * 销量
df['total'] = df['price'] * df['sale']
print(df)

# 找出销售额最高的产品
print(df.sort_values('total', ascending=False).head(1))
print(df.nlargest(1, 'total'))

# 按销售额从低到高排序
print(df.sort_values('total'))

pass
```

```python
data = {
    'id': [101,102,103,104,105],
    'name': ['A', 'B', 'C', 'D', 'E'],
    'type':['电子', '服饰','电子', '居家', '服饰'],
    'price': [1200,300,800,150,200],
    'number': [1,3,2,5,4]
}
df = pd.DataFrame(data)
print(df)
# 计算每位用户的销售总额
df['total'] = df['price'] * df['number']
print(df)
# 消费金额最高的用户
print(df.nlargest(1, 'total'))
# 计算所有用户的平均消费金额
print(df['total'].sum())
print(df['total'].mean())
# 统计电子产品的总购买数量
print(df[df['type'] == '电子'])
print(df[df['type'] == '电子']['number'])
print(df[df['type'] == '电子']['number'].sum())
pass
```
