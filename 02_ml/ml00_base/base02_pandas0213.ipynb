```python
# 1.2 缺失数据的处理
# nan: not a number
print(None, type(None))
print(np.nan, type(np.nan))
print(pd.NA, type(pd.NA))

s = pd.Series([2,None, 3, np.nan, 5, pd.NA])
print(s)
df = pd.DataFrame([
    ['张三', None, '王五', np.nan, '钱七', pd.NA],
    [18, 28, None, 26, np.nan, pd.NA],
    ['第1列', '第2列', '第3列', '第4列', '第5列', '第6列']
])
df.columns = ['第1列', '第2列', '第3列', '第4列', '第5列', '第6列']
print(df)

# 如何检测是否存在缺失值
print(s.isna())
print(s.isnull())
print(s.isnull().sum())
print(df.isna())
print(df.isnull())
print(df.isna().sum())
print(df.isnull().sum())
print(df.isnull().sum(axis=1))
print()

# 缺失值处理，直接剔除存在缺失值的数据 或 用默认值填充缺失值
# 剔除
print(s)
print(s.dropna())
print()
# dataFrame按行剔除数据，一行就是一条数据，只要这条数据里面出现了缺失值那么这条数据也就没用了，所有是按行剔除数据。【行是样本，列是标签】
print(df)
print(df.dropna())
# 指定列删除缺失值
# print(df.dropna(subset=['users']))

# 填充默认值
df = pd.read_csv('../../data/ml/base/weather_withna.csv')
print(df)
# 查看每一行都有几个缺失值
print(df.isna().sum(axis=0))
# 查看每一列都有几个缺失值
print(df.isna().sum(axis=0))
# precipitation 降雨量， 用平均值填充
# 1. 获取precipitation这一列所有数据
print(df.loc[:,'precipitation'])
# 2.找到precipitation这一列有缺失的数据
print(df.loc[df.loc[:,'precipitation'].isna(),'precipitation'])
# 3.precipitation获取这一列的平均值
print(df.loc[:,'precipitation'].mean())
# 4.precipitation将这一列是缺失的用平均值回填
df.loc[df.loc[:,'precipitation'].isna(),'precipitation'] = df['precipitation'].mean()
# 5.输出新的数据
print(df.loc[:,'precipitation'])

# 使用pd提供的方法，df.fillna, 该方法只能对nan的数据生效，如果要填充的那一列不存在nan数据了，那么该函数执行后没有修改任何数据，看起来没效果
print(df.fillna({'precipitation': 2 + 1}))
print(df.fillna({'temp_max': df['temp_max'].max()}))
# 同时补齐多个列
print(df.fillna(df[['temp_min', 'wind']].mean()))
# 使用缺失值上下的数据进行填充, ffill,bfill   front, back
print(df.ffill())

# 查看每一列都有几个缺失值
print(df.isna().sum(axis=0))
pass
```

```python
# 1.5 数据变形
# pd.melt(),把宽表转化为长表
# pd.pivot(),把长表转化为宽表
data = {
    'id': [1,2,3],
    'name': ['A', 'B', 'C'],
    'math':[85,95,45],
    'english':[85,85,65],
    'chinese':[75,95,95],
}
df = pd.DataFrame(data)
print(df)
# id_vars 表示哪些列不需要拆分
pd2 = pd.melt(df, id_vars=['id', 'name'])
pd3 = pd.melt(df, id_vars=['id', 'name'], var_name='科目', value_name='成绩')
print(pd2)
print(pd3)

# index表示不变的列
df5 = pd.pivot(pd3, index=['id', 'name'], columns='科目', values='成绩')
print(df5)
print(df5.shape)
```

```python
# 分组聚合
# df.groupby('分组字段')['聚合字段'].聚合函数()

df = pd.read_csv('../../data/ml/base/employees.csv')
# print(df)
print(df['department_id'].isna().sum())
df = df.dropna(subset=['department_id'])
print(df['department_id'].isna().sum())
# 对 department_id 这一列的float64转化为int64
df['department_id'] = df['department_id'].astype('int64')
print(df.info())
# 按部门进行分组, 求每一个部门的薪资情况
print(df.groupby('department_id').groups)
print(df.groupby('department_id')['salary'].sum())
# 按不同部门不同岗位进行分组, 求每一个部门的薪资情况
print(df.groupby(by=['department_id', 'job_id']).groups)
print(df.groupby(by=['department_id', 'job_id'])['salary'].sum())
print(df.groupby(by=['department_id', 'job_id'])['salary'].sum())
```

```python
from dataclasses import dataclass
from collections import defaultdict

@dataclass
class Student:
    school_id: int
    class_id: int
    student_id: int
    name: str

students = []
students.append(Student(101,1011,101101,'张三101101'))
students.append(Student(101,1011,101102,'张三101102'))
students.append(Student(101,1012,101201,'张三101201'))
students.append(Student(101,1012,101202,'张三101202'))
students.append(Student(102,1021,102101,'张三102101'))
students.append(Student(102,1021,102102,'张三102102'))
print(students)

# 我现在要把学生的学校,班级存储到dict的key中便于快速查找对应学校,班级的数据,以下我实现了两个方法,请对以下两个方法进行评价

print("\n方法1")
student_map1 = defaultdict(list)
for student in students:
    student_map1[f"{student.school_id}-{student.class_id}"].append(student)
print(student_map1)

print("\n方法2")
student_map2 = defaultdict(list)
for student in students:
    student_map2[(student.school_id, student.class_id)].append(student)
print(student_map2)
print()
print(student_map2.get((101, 1011)))
print(student_map2.get((101, 1011))[0])
print(student_map2.get((101, 1011))[1])


```
