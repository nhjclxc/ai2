```python
import numpy as np
import pandas as pd
# import matplotlib
import matplotlib.pyplot as plt
print(np.__version__)
print(pd.__version__)
# print(matplotlib.__version__)
# 显示中文问题
plt.rcParams['font.sans-serif'] = ['SimHei']      # 使用黑体
plt.rcParams['axes.unicode_minus'] = False        # 解决负号显示问题
```

# 1. 折线图

```python

# 1. 导入matplotlib库
# 2. 创建图表
# 3. 数据准备
# 4. 绘制图表
# 5. 显示图表

# 2. 创建图表
plt.figure(figsize=(10,5)) # 可以理解为这个是纸张大小

# 3. 数据准备
month = ['1月', '2月', '3月', '4月', '5月', '6月', '7月', '8月', '9月', '10月', '11月', '12月']
np.random.seed(123)
sales_a = np.random.randint(50,100,size=12)
sales_b = np.random.randint(50,100,size=12)

# 4. 绘制图表
plt.plot(month, list(sales_a), label='产品A')
plt.plot(month, list(sales_b), label='产品B')
# 添加标题
plt.title('2025年销售额', color='red', fontsize=23)
# 添加x轴标签
plt.xlabel('月份', fontsize=15)
# 添加y轴标签
plt.ylabel('销售额(万元)', fontsize=15)
# 添加图列
plt.legend(loc = 'upper left')
# 添加网格线
# plt.grid(True)
# plt.grid(axis='x')
plt.grid(axis='y', linestyle='--', alpha=0.3, color='red')
# 设置y轴范围
plt.ylim(30, 120)
# 设置数据标记(点对应的数值显示出来)
for x,ya,yb in zip(month, sales_a, sales_b):
    # 前面两个参数是显示的文本先对于那个点的偏移
    plt.text(x,ya+5, f"({x}, {ya})")
# 5. 显示图表
plt.show()
```
